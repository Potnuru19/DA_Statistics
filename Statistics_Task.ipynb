1. Course (DA, PFS or JFS): Nominal categorical - because the categories have no natural order.

2. Satisfaction (Dissatisfied, Neutral or Satisfied): Ordinal categorical - because the categories have a meaningful order.

3. Assignments completed (0,1,2,...): Discrete numerical - because it is counted in whole numbers.

4. Task completion time (measured in minutes, including fractions): Continuous numerical - because it can take any value, including decimal values.

(B) Explain why the mean of student IDs does not describe a useful student characteristic.

Student IDs are identifiers used only to distinguish students. They are nominal data and do not represent measurable quantities.
Therefore, calculating their mean has no meaningful interpretation because the numerical values are arbitrary and do not represent any actual student characteristic.
